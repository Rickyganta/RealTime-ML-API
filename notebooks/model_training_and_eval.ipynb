# Movie Recommendation Training + Offline Evaluation

This notebook demonstrates MovieLens loading, SVD training, and NDCG evaluation.

```python
import pandas as pd
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split
from sklearn.metrics import ndcg_score
import numpy as np
```

```python
ratings = pd.read_csv('../data/ml-latest-small/ratings.csv')
reader = Reader(rating_scale=(0.5, 5.0))
data = Dataset.load_from_df(ratings[['userId','movieId','rating']], reader)
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)
model = SVD(random_state=42)
model.fit(trainset)
preds = model.test(testset)
rmse = np.sqrt(np.mean([(p.est - p.r_ui)**2 for p in preds]))
rmse
```

```python
# Simplified ranking metric example
y_true = np.array([[3, 2, 3, 0, 1, 2]])
y_score = np.array([[0.9, 0.8, 0.7, 0.1, 0.3, 0.2]])
ndcg_score(y_true, y_score)
```
